```python
with open("6LU7.pdb", "r") as f_in, open("6lu7_final.pdb", "w") as f_out:
    for line in f_in:
        # 只保留蛋白 A 链
        if line.startswith("ATOM") and line[21] == 'A':
            f_out.write(line)
        # 将 C 链的所有记录强制转为 HETATM 02J
        elif line[21] == 'C' and (line.startswith("ATOM") or line.startswith("HETATM")):
            new_line = "HETATM" + line[6:17] + "02J" + line[20:]
            f_out.write(new_line)
        elif line.startswith("TER") and line[21] == 'A':
            f_out.write(line)
```
